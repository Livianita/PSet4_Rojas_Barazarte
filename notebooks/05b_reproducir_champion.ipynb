# 05b · Reproducir el champion del PSet 3 desde el Model Registry

Notebook limpio: no depende de ninguna variable de los anteriores. Carga la versión 1 de
`favorita-perecederos-modelo` (Chronos cuantil), el champion elegido en el PSet 3, reconstruye el conjunto de prueba
con el mismo código y la misma muestra (huella SHA-256 verificada) y comprueba que el RMSLE y el costo de negocio de
la prueba (Co:Cu 1:3) son **exactamente** los registrados en su corrida final. La salida del modelo es el pedido
sugerido: el costo se calcula directo sobre ella.

La versión queda fija y no se toma del alias `@champion`. En el PSet 4 el modelo de producción pasó a GB cuantil
(versión 3, ver `docs/cambios.md`) y durante la simulación el alias pasó a los reentrenamientos aprobados (v7 y luego
v8), que no se evaluaron sobre esta prueba. El champion vigente y sus reentrenamientos se analizan en
`06_simulacion_produccion.ipynb`.

```python
import os
import sys

sys.path.insert(0, '..')
os.environ.setdefault('MLFLOW_DISABLE_AGENT_HINT', '1')
os.environ.setdefault('MLFLOW_ENABLE_ARTIFACTS_PROGRESS_BAR', 'false')
os.environ.setdefault('TQDM_DISABLE', '1')
os.environ['MLFLOW_ALLOW_MODEL_CODE_EXECUTION'] = 'true'    # modelo del servidor de tracking del proyecto (formato skops)

import mlflow
import numpy as np
from dotenv import load_dotenv
from mlflow.tracking import MlflowClient

from src.datos import cargar_muestra
from src.pipeline import SERIE, construir_features, dividir, rmsle

load_dotenv('../.env')
VERSION_PSET3 = '1'                       # champion del PSet 3; el alias @champion ya apunta a otra versión
URI = f'models:/favorita-perecederos-modelo/{VERSION_PSET3}'
cliente = MlflowClient()
version = cliente.get_model_version('favorita-perecederos-modelo', VERSION_PSET3)
registradas = cliente.get_run(version.run_id).data.metrics
print(f'{URI} → versión {version.version}, run_id {version.run_id}, modelo {version.tags.get("modelo")}')
print(f"Registrado: RMSLE {registradas['rmsle_prueba']!r} · costo 1:3 {registradas['costo_1_3']!r}")
```

```python
modelo = mlflow.sklearn.load_model(URI)
ventas, meta = cargar_muestra()
_, X_prueba, _, y_prueba = dividir(construir_features(ventas), columnas_extra=SERIE + ['DATE'])
pedido = np.clip(modelo.predict(X_prueba), 0, None)
y = y_prueba.to_numpy(dtype=float)
reproducidas = {'rmsle_prueba': rmsle(y, pedido),
                'costo_1_3': 1 * np.maximum(pedido - y, 0).sum() + 3 * np.maximum(y - pedido, 0).sum()}
print(f'Huella de la muestra: {meta["sha256"]}')
for nombre, valor in reproducidas.items():
    exacta = np.isclose(valor, registradas[nombre], rtol=1e-12, atol=1e-12)
    print(f'{nombre}: reproducido {valor!r} · diferencia {abs(valor - registradas[nombre]):.2e} → '
          f'{"OK" if exacta else "revisa"}')
```
